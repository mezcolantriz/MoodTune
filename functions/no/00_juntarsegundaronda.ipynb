```python
import pandas as pd

# Rutas de los archivos
nuevo_path = r"C:\Users\solan\MoodTune\data\procesando\df_80-200_p4_last.csv"
traducidas_path = r"C:\Users\solan\MoodTune\data\procesando\df_80-200_p4_last_traducidas.csv"
final_path = r"C:\Users\solan\MoodTune\data\procesando\df_80-200_final_1.csv"
output_path = r"C:\Users\solan\MoodTune\data\procesando\df_80-200_final_1.csv"

# Cargar archivos
df_nuevo = pd.read_csv(nuevo_path)
df_traducidas = pd.read_csv(traducidas_path)
df_final = pd.read_csv(final_path)

# Unir traducidas con los nuevos datos (por recording_id)
df_nuevo = df_nuevo.set_index('recording_id')
df_traducidas = df_traducidas.set_index('recording_id')

# Añadir translated_lyrics y language si están en traducidas
if 'translated_lyrics' in df_traducidas.columns:
    df_nuevo['translated_lyrics'] = df_traducidas['translated_lyrics']

if 'language' in df_traducidas.columns:
    df_nuevo['language'] = df_traducidas['language']

df_nuevo.reset_index(inplace=True)

# Concatenar el nuevo bloque de datos al final del final existente
df_final_actualizado = pd.concat([df_final, df_nuevo], ignore_index=True)

# Guardar el archivo combinado
df_final_actualizado.to_csv(output_path, index=False)

print(f"✅ Nuevas canciones añadidas y archivo actualizado en: {output_path}")
```

```python
path_chequeo = r"C:\Users\solan\MoodTune\data\procesando\df_80-200_p4.csv"
df_chequeo = pd.read_csv(path_chequeo)
# nulos en language
df_chequeo['language'].isnull().sum()
```

```python
def rellenar_language(row):
    if pd.isna(row['language']):
        resultado = df_chequeo[df_chequeo['recording_id'] == row['recording_id']]
        if not resultado.empty:
            return resultado['language'].values[0]
    return row['language']

df_final_actualizado['language'] = df_final_actualizado.apply(rellenar_language, axis=1)
```
